```python
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedGroupKFold, GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
import m2cgen as m2c
```

# Binary Sprint Classes Model


```python
df = pd.read_csv("../data/features/engineered_features_v1.csv")
```

```python
X = df.drop(columns=["trial_num", "start", "end", "sprint"])
y = df["sprint"]
groups = df["trial_num"]
sgkf = StratifiedGroupKFold(n_splits=4)
```

```python
importances_list = []

for train_idx, test_idx in sgkf.split(X, y, groups=groups):
    X_train = X.iloc[train_idx]
    y_train = y.iloc[train_idx]
    X_test = X.iloc[test_idx]
    y_test = y.iloc[test_idx]

    clf = RandomForestClassifier(random_state=1)
    clf.fit(X_train, y_train)

    importances_list.append(clf.feature_importances_)
    print(classification_report(y_test, clf.predict(X_test)))

importances_df = pd.DataFrame(importances_list, columns=X.columns).T

for col in importances_df.columns:
    importances_df[f"{col}_above_baseline"] = importances_df[col].apply(
        lambda x: "yes" if x > 1 / len(importances_df) else "no"
    )
importances_df
```

```python
passing_features = ['clean_accel_rms', 'clean_accel_mean', 'raw_accel_rms', 'raw_accel_std', 'raw_accel_mean', 'clean_accel_std', 'clean_accel_max', 'raw_accel_max']
X = df.loc[:, passing_features]
X.corr()
```

```python
selected_features = ["clean_accel_rms", "clean_accel_max", "clean_accel_std"]
X = df[selected_features]
```

```python
depths = list(range(1, 10, 1))
estimators = list(range(70, 90, 2))

param_grid = {"max_depth": depths, "n_estimators": estimators}

grid_search = GridSearchCV(
    estimator=RandomForestClassifier(random_state=1),
    param_grid=param_grid,
    cv=sgkf,
    scoring="accuracy",
    n_jobs=-1,
)

grid_search.fit(X, y, groups=groups)
```

```python
clf = RandomForestClassifier(max_depth=2, n_estimators=70, random_state=1)
clf.fit(X, y)
```

```python
model_src = m2c.export_to_dart(clf)

with open("../models/binary_sprint_classifier.dart", "w") as file:
    file.write(model_src)
```
